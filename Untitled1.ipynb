```python
#chi square test categorical vs categorical varible me apply hota hai
```

```python
categorical_variables = ['Region_Europe','Region_RestOfWorld','Model_5 Series','Model_MINI','Model_X3','Model_X5','Model_X7','Model_i4','Model_iX','GDP_Growth_good','GDP_Growth_excellent','GDP_Growth_brilliant'
]
```

```python
#to make the exam score as categorical varible using create a bins
```

```python
from scipy.stats import chi2_contingency
import pandas as pd

alpha = 0.05

# Create 4 bins
dataclean['Fuel_Price_Index_bin'] = pd.qcut(dataclean['Fuel_Price_Index'], q=4, labels=False)

chi2_results = {}

for col in categorical_variables:
    contingency = pd.crosstab(dataclean[col], dataclean['Fuel_Price_Index_bin'])
    chi2_stat, p_val, _, _ = chi2_contingency(contingency)

    decision = 'Reject Null (Keep Feature)' if p_val < alpha else 'Accept Null (Drop Feature)'

    chi2_results[col] = {
        'chi2_statistic': chi2_stat,
        'p_value': p_val,
        'Decision': decision
    }

# 👇 OUTSIDE LOOP
chi2_data = pd.DataFrame(chi2_results).T
chi2_data = chi2_data.sort_values(by='p_value')

chi2_data
```
